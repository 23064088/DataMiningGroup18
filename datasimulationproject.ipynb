```python
!pip install -q sdv pandas numpy

import pandas as pd
import numpy as np
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import SingleTableMetadata

#Create small dataset manually to teach the model
seed_data = {
    'Customer_ID': range(1001, 1051), #50 row
    'Age': np.random.choice([25, 30, 35, 45, 50, 60, np.nan], 50),
    'Income': np.random.normal(55000, 15000, 50).astype(int),
    'Credit_Score': np.random.randint(600, 850, 50),
    'Investment_Returns': np.random.uniform(2, 12, 50).round(2),
    'Risk_Level': np.random.choice(['Low', 'Medium', 'High'], 50),
    'Customer_Feedback': np.random.choice([
        "Great service!", "Average.", "Fees are high.",
        "Happy with the app.", "The app is [MASK].", np.nan
    ], 50)
}

df_train = pd.DataFrame(seed_data)

#CTGAN model
meta = SingleTableMetadata()
meta.detect_from_dataframe(df_train)

print("Training the model... (this might take a moment)")
ctgan = CTGANSynthesizer(meta)
ctgan.fit(df_train)

#Generate the actual 1000 rows
print("Generating new data...")
new_data = ctgan.sample(num_rows=1000)

#Create new ids that are distinct
new_data['Customer_ID'] = range(1001, 2001)

# Save to CSV
new_data.to_csv('financial_data.csv', index=False)

print("Done. 'financial_data.csv' is ready.")
print(new_data.head())

```

```python
from transformers import pipeline
import pandas as pd
import numpy as np
```

```python
#Load the Small Language Model (SLM)
print("Loading BERT-tiny for text imputation")
fill_mask = pipeline('fill-mask', model='prajjwal1/bert-tiny')
```

```python
#Define the Imputation Logic
def slm_impute_feedback(text):
    """
    If text is NaN or contains '[MASK]', use the AI to fill it.
    """
    # When missing (NaN)
    if pd.isna(text):
        #prompt to generarte new word
        prompt = "The service was [MASK]."
        prediction = fill_mask(prompt)[0]['token_str']
        return f"The service was {prediction}."

    #Have [MASK]
    if "[MASK]" in str(text):
        prediction = fill_mask(text)[0]['token_str']
        return text.replace("[MASK]", prediction)

    #Text is fine
    return text
```

```python
#Apply to the Dataset
if 'df' not in locals():
    df = pd.read_csv('financial_data.csv') #make sure fetch the latest one

print("\nImputing Missing Text")

print("Before:", df['Customer_Feedback'].tail(5).values)

df['Customer_Feedback'] = df['Customer_Feedback'].apply(slm_impute_feedback)

print("After: ", df['Customer_Feedback'].tail(5).values)

# Save the cleaned data
df.to_csv('financial_data.csv', index=False)
print("✅ Text Imputation Complete. Data saved.")
```

```python
#Numerical Preprocessing
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np

df = pd.read_csv('financial_data.csv')

print("\nPreprocessing Status")
print(f"Missing Age values before: {df['Age'].isna().sum()}")
```

```python
#Impute Missing Values
print("Filling missing Age and Income with Median ...")
imputer = SimpleImputer(strategy='median')

# Apply imputation to Age and Income columns
cols_to_fix = ['Age', 'Income']
df[cols_to_fix] = imputer.fit_transform(df[cols_to_fix])

print(f"Missing Age values after: {df['Age'].isna().sum()}")
```

```python
#Normalize Features
# We scale Income and Credit Score to be between 0 and 1
# This is crucial for models like Neural Networks and KNN
print("Scaling Income and Credit Score to 0-1 range ...")
scaler = MinMaxScaler()

cols_to_scale = ['Income', 'Credit_Score']
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
```

```python
#Save the processed data
df.to_csv('financial_data.csv', index=False)

print("\nSample of Processed Data:")
print(df[['Age', 'Income', 'Credit_Score']].head())
print("Numerical Preprocessing Complete. Data saved.")
```
